# AI-Based Smart Attendance Management System

## Project Overview
This project automates student attendance using face recognition, a liveness check, SQLite database storage, and a Streamlit dashboard.

## Main Features
- Face detection and recognition using InsightFace/ArcFace
- Basic liveness verification using head movement
- Student registration and identity management
- SQLite storage for attendance records
- Class-hour attendance tracking
- Student-wise attendance summary and percentage calculation
- Dashboard filters and CSV export

## Technologies Used
- Python
- OpenCV
- InsightFace
- NumPy
- SQLite
- Streamlit
- CSV

## Project Files
- `webcam_recognition (1).py` — webcam recognition and attendance workflow
- `dashboard.py` — Streamlit dashboard
- `database.py` — database-related functions
- `arcface_embeddings.npy` — face embeddings
- `arcface_names.npy` — names associated with embeddings
- `attendance.csv` — attendance CSV file
- `attendance.db` — SQLite database (stored in the configured database path)

## Setup
1. Install Python and the packages required by the project.
2. Install the appropriate InsightFace, ONNX Runtime, OpenCV, NumPy and Streamlit dependencies.
3. Ensure the face embedding files and SQLite database are in their configured locations.
4. Connect a webcam.
5. Run the recognition script to test attendance recording.
6. Start the dashboard with:

   `python -m streamlit run dashboard.py --server.port 8501`

7. Open `http://localhost:8501` in a browser.

## Database
The SQLite database contains registered student details and attendance records. Back up the database before making changes.

## Limitations
The current liveness mechanism uses basic head-movement checks and is not a complete anti-spoofing solution. Recognition accuracy depends on lighting, camera quality, face images and threshold configuration.
